```octave
[~, genotype] = xlsread ('../data/abeta-cleaned_ver1_coded.xlsx', 'Abeta', 'E2:E199');
[~, condition] = xlsread ('../data/abeta-cleaned_ver1_coded.xlsx', 'Abeta', 'F2:F199');

% for cell clusters am I averaging from the same cell into one data point
% need to load animal information once it is available

outcomes = xlsread ('../data/abeta-cleaned_ver1_coded.xlsx', 'Abeta', 'G2:S199');

disp (size(genotype))
disp (size(condition))
disp (size(outcomes))
```

```octave
% getting rid of columns now as matrix automatically removes NaN, remember that now means that the outcome numbers have shifted 
```

```octave
if (size(outcomes, 2) > 11)
    outcomes(:,[8,13]) = [];
end

disp (size(outcomes))
```

```octave
for i = 1: size(outcomes, 2)
    MAT_T = bootlm (outcomes(:,i), {genotype, condition}, 'model', 'full', 'display', 'on', ...
            'contrasts', 'treatment', 'nboot', 0);
    print(sprintf('../output/NC_ver1/outcome%d.png', i), '-dpng', '-r300')
end
```

```octave
% after looking at graphs, outlier rows are: row 2 (outcome 7)  row 164 (outcome 11) (double check)
```

```octave
if (size (outcomes, 1) > 196)
    genotype([2,164],:) = [];
    condition([2,164],:) = [];
    outcomes([2,164],:) = [];
    MAT_T.X([2,164],:) = [];
end

disp (size(genotype))
disp (size(condition))
disp (size(outcomes))
disp (size(MAT_T.X))
```

```octave
% bayesian bootstrap summary statistics using Simple Random Sampling
```

```octave
STATS_SRS = bootbayes(outcomes, MAT_T.X);
```
