```octave
# Load the predictor genotype - it has two levels: G1 and G2
[~, genotype]  = xlsread ('../data/basic_properties_data_coded.xlsx', 'OGS', 'E2:E77');

# Load the predictor condition - it has three levels: C1, C2 and C3
[~, condition] = xlsread ('../data/basic_properties_data_coded.xlsx', 'OGS', 'F2:F77');

# Load the animal cluster ids
animal = xlsread ('../data/basic_properties_data_coded.xlsx', 'OGS', 'B2:B77');

# Fetch matrix of all the outcome measures
outcomes = xlsread ('../data/basic_properties_data_coded.xlsx', 'OGS', 'G2:R77');

# Get size of the data
disp(size(genotype))
disp(size(condition))
disp(size(animal))
disp(size(outcomes))

# Note that outcomes has 12 columns: a different outcome measure for each of the 76 rows (cell recordings) 
```

```octave
# Fit multiple univariate linear models by ordinary least squares to each of the outcomes and save the diagnostic plots
for i = 1:size(outcomes, 2)

  # By default bootlm assumes variables listed in cell array of second input argument are categorical.
  # Set 'model' to 'full' so that it includes the interaction.
  # Set 'nboot' to 0 to bypass bootstrap resampling.
  # Set 'display' to 'on' to create diagnostic plots.
  # Set 'contrasts' to 'treatment' (dummy) coding
  MAT_T = bootlm (outcomes(:,i), {genotype, condition}, 'model', 'full', 'nboot', 0, 'display', 'on', ...
                  'contrasts', 'treatment');

  # Save the figure of the diagnostic plots
  print (sprintf('../output/Demo/outcome%d.png', i),'-dpng','-r300')
  
end

# Diagnostic plots shown in the output below are just for the last column.
# Figures of the diagnostic plots have been saved.
```

```octave
if (size (outcomes, 1) > 75)

  # Remove outlier (row 6) from the original data
  genotype(6,:) = [];
  condition(6,:) = [];
  animal(6,:) = []; 
  outcomes(6,:) = [];

  # ... and the design matrix (X) in MAT_T
  MAT_T.X(6,:) = [];

end

# Get size of the data now - all should have 75 rows) 
disp(size(genotype))
disp(size(condition))
disp(size(animal))
disp(size(outcomes))
disp(size(MAT_T.X))
```

```octave
# Get bayesian bootstrap summary statistics of all outcomes (Simple Random Sampling)
STATS_SRS = bootbayes (outcomes, MAT_T.X);
```

```octave
# Get bayesian cluster bootstrap summary statistics
STATS = bootbayes (outcomes, MAT_T.X, animal);
```

```octave
# Estimate design effect per outcome
Deff_per_outcome = mean (STATS.stdev.^2 ./ STATS_SRS.stdev.^2, 1);
Deff = mean (Deff_per_outcome);

disp (Deff_per_outcome)
disp (Deff)
m = harmmean (accumarray (animal, 1, [], @sum));  % m is average cluster size
ICC = (Deff - 1) / (m - 1);                       % Intraclass correlation coefficient (rho)
disp (ICC)
```

```octave
# Get design matrix for orthogonal (helmert) contrast coding and hypothesis matrix for estimated marginal means
# This time we set display off.
MAT_H = bootlm (outcomes(:,11), {genotype, condition}, 'model', 'full', 'nboot', 0, 'display', 'off', ...
                'contrasts', 'helmert', 'dim', [1, 2]);
MAT_H.C
MAT_H = bootlm (outcomes(:,11), {genotype, condition}, 'model', 'full', 'nboot', 0, 'display', 'off', ...
                'contrasts', MAT_H.C, 'dim', [1, 2]);
# Get dimenions of the design matrix (which was constructed using sign-flipped helmert contrasts)
size (MAT_H.X)
# It has 75 rows (cell recordings) and 6 columns (1 intercept and 5 predictor terms, all categorical)
```

https://gnu-octave.github.io/statistics-resampling/function/bootridge.html

```octave
# Lets fit using ridge regression and interpret the coefficients coded with sign-flipped helmert contrasts.
# -- Function File: bootridge (Y, X, CATEGOR, NBOOT, ALPHA, L, DEFF, SEED, TOL)
# Note that we specify CATEGOR input argument as vector 2:6 so that columns 2:6 are treated as categorical
# and we supply our calculated estimate of the design effect.
tic
bootridge(outcomes, MAT_H.X, '*', 1999, .05, [], Deff);
toc
ridge_coeff = ans;
```

```octave
% Stability selection probabilities of the regression coefficients
disp(ridge_coeff.stability*100)
```

## Interpreting the coefficients from sign-flipped helmert contrasts
1. is the intercept so represents the grand mean  
2. is the difference between the two genotypes (G2 - G1) averaged over the conditions  
3. is the difference between conditions (C2C3 - C1) averaged over the genotypes  
4. is the difference between conditions (C3 - C2) averaged over the genotypes  
5. is the difference between the two genotypes (G2 - G1) different in C2C3 compared to C1?  
6. is the difference between the two genotypes (G2 - G1) different in C3 compared to C2?  

The outcome numbers correspond to.
1. peak 
2. afterhyp 
3. amp 
4. thresh 
5. half (full-width half-maximum)
6. rise 
7. fall 
8. IR (Input resistance)
9. Rh (Rheobase)
10. sag (Ih sag)
11. RMP (resting membrane potential)
12. maxApnum 

Things I have noticed - there is strong evidence  (lnBF10 = +9.51, BF10 = 13494, SS = 99.9%) for a 3.85 mV decrease in Ih sag, strong evidence for a 0.09 ms narrowing of the action potential (full-width half-maximum) (lnBF10 = +2.00, BF10 = 7.39, SS = 99.2%), and strong evidence for a 1.76 Hz increase in maximum AP firing frequency (lnBF10 = +2.21, BF10 = 9.12, SS = 99.7%), with ketamine/memantine irrespective of genotype. For AP firing frequency, there is some (anecdotal) evidence to indicate that that the genotype effect dependence on drug is a strong predictor of maximum firing rate (lnBF10 = +0.79, BF10 = 2.21, SS = 98.9%). There is good evidence for G2 having an 17.9 Mohm difference between the genotypes depending on the drug condition C1 vs C2C3 (lnBF10 = +1.47, BF10 = 4.35, SS = 99.9%). There is also some anecdotal evidence for a 2.05 mV increase in AP firing threshold in C2C3 compared to C1 irrespective of genotype (lnBF10 = +0.732, BF10 = 2.08, SS = 96.0%).

There are also some effects on afterhyperpolarization and resting membrane potential... can you find and describe them Nimjeet?


```octave
# Get the correlations for any pairs of outcomes involving column 11 of outcomes (Apnum)
RTAB = ridge_coeff.RTAB;
disp(min(RTAB(:,3)));
disp(max(RTAB(:,3)));
contains_Apnum = any (RTAB(:,1:2) == 12, 2);
RTAB(contains_Apnum,:)
# Columns 1 and 2 are the column numbers of the matrix of outcomes for each pair
# Column 3 is the correlation coefficient between the pair of outcomes.
# Columns 4 and 5 are the lower and upper 95% credible intervals (flat prior)
```

## Notes

The outcome numbers correspond to.
1. peak 
2. afterhyp 
3. amp 
4. thresh 
5. half (full-width half-maximum)
6. rise 
7. fall 
8. IR (Input resistance)
9. Rh (Rheobase)
10. sag (Ih sag)
11. RMP (resting membrane potential)
12. maxApnum  

Pair-wise correlations between outcomes ranged from -.75 to +.79. With reference to maximum action potential firing rate, a negative correlation was found with rheobase (-.38), repolarization rate (-.50), and full-width half-maximum (-.37). Therefore, lower rheobase, faster AP repolarization and shorter AP full-width half-maximum are associated with higher maximum AP firing frequency. In contrast, maximum action potential firing frequency was positively correlated with AP amplitude (+.40), peak voltage (+.39) and depolarization rate (+.41). Therefore, larger amplitude, peak and faster rise of APs are associated with higher maximum AP firing frequency.

```octave
# Lets fit using ridge regression and use the hypothesis matrix to return the group estimated marginal means.
# -- Function File: bootridge (Y, X, CATEGOR, NBOOT, ALPHA, L, DEFF, SEED, TOL)
# Note that we specify CATEGOR input argument as vector 2:6 so that columns 2:6 are treated as categorical
# and we supply our calculated estimate of the design effect.
tic
bootridge(outcomes, MAT_H.X, "*", 1999, .05, MAT_H.L, Deff);
toc
ridge_emm = ans;

# Note that we expect NaN for Bayes factors (BF10) and U (-Inf, Inf) for the prior on all the linear estimates
# of the marginal means since they all involve the intercept.
```

```octave
# Now lets fit the data again to get the hypothesis matrix for specific group comparisons we are interested in.
comparisons = [2 1; 4 3; 6 5; 3 1; 5 1; 5 3; 4 2; 6 2; 6 4]

# Get design matrix hypothesis matrix for estimated marginal means. We could use any outcome but we'll pick 11.
# This time we set display off. 
MAT_H_PAIRS = bootlm (outcomes(:,11), {genotype, condition}, 'model', 'full', 'nboot', 0, 'display', 'off', ...
                      'contrasts', 'helmert', 'dim', [1, 2], 'posthoc', comparisons);
```

```octave
# Lets fit using ridge regression and use the hypothesis matrix to return the comparisons of interest.
# -- Function File: bootridge (Y, X, CATEGOR, NBOOT, ALPHA, L, DEFF, SEED, TOL)
# Note that we specify CATEGOR input argument as vector 2:6 so that columns 2:6 are treated as categorical
# and we supply our calculated estimate of the design effect.
tic
bootridge(outcomes, MAT_H_PAIRS.X, [2:6], 1999, .05, MAT_H_PAIRS.L, Deff);
toc
ridge_pairs = ans;
# Note that outcome 8 is input resistance
```
